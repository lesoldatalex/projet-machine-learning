# 3 : Formalisation of the problem 

The objective of our project is to recognize the presence and position of humans in images of landscapes.
In other words, we want to build a model capable of identifying whether a human appears in an image, and in which position or category (for example: standing, sitting, lying down, or no human present).

This makes it a multi-class image classification problem.



The objective of this project is to **detect the presence and location of humans within landscape images**. In other words, we aim to develop a model capable of identifying **where humans appear in an image**, by providing bounding boxes or coordinates around detected individuals. This method is called labelisation.

This corresponds to a **supervised object detection problem** in computer vision.

### Definition of the Problem  
Given an input image $X \in \mathbb{R}^{H \times W \times 3}$,  
the goal is to learn a function:

$$
f_\theta(X) = \{(b_i, c_i)\}_{i=1}^{N}
$$

where each \( b_i \) represents the bounding box coordinates of a detected object (in this case, a human), and \( c_i \) represents the predicted class label (“human”).  

The model must therefore predict **both**:
1. The **class** of the object (here, human vs background).  
2. The **location** of each detected object within the image.

The training data consists of images annotated with bounding boxes that indicate the positions of humans (standing,laying down, sitting). This allows the model to learn visual patterns such as body shapes, edges, textures, and contextual cues that distinguish humans from other landscape elements like trees, rocks, or buildings.

### Learning Framework  
The task is framed as a **supervised learning** problem:
- **Inputs:** labeled images containing one or several humans within various landscapes.  
- **Outputs:** coordinates and class labels of detected humans.  
- **Loss function:** typically a combination of a **classification loss** (to recognize humans) and a **localization loss** (to predict bounding box positions), as used in models like YOLO, Faster R-CNN, or SSD.

### Relevance of the Dataset  
Our dataset is appropriate for this problem because it contains:
- **Diverse environmental conditions** (lighting, perspective, backgrounds).  
- **Variations in human size, posture, and position** within the scene.  
- **Realistic visual complexity**, making the detection task representativ  e of real-world applications.

This diversity supports robust learning and helps the model generalize to new, unseen images.

### Expected Outcome  
The final goal is to train a model capable of **automatically detecting and localizing humans** in landscape images. The system will serve as a foundation for further computer vision tasks, such as human tracking, behavior analysis, or automatic content filtering.






# Part 4 — Selection of a Baseline Model and Implementation

The purpose of this section is to establish a **baseline performance** for our human detection task before applying any fine-tuning or advanced optimization.  
We selected two complementary baseline models:

1. A **Convolutional Neural Network (CNN)** implemented from scratch, used as a simple starting point for visual feature extraction.  
2. A **YOLO (You Only Look Once)** object detection model, applied without fine-tuning, using pre-trained weights on a general dataset such as COCO.

---

### 4.1. Baseline Model 1 — Simple CNN

A simple CNN was designed to classify whether an image contains a human or not, based on the presence of relevant visual patterns.

#### **Architecture**

The CNN architecture consists of several convolutional layers that extract visual features from the input images, each followed by ReLU activation functions and max-pooling operations to reduce spatial dimensions while preserving essential information. After the convolutional blocks, the output feature maps are flattened and passed through fully connected layers that learn higher-level representations. Finally, a sigmoid or softmax output layer is used, depending on whether the task is binary or multi-class classification.

#### **Training Configuration**
- **Input size:** 128 × 128 RGB images  
- **Optimizer:** Adam  
- **Loss function:** Binary Cross-Entropy (for human vs no-human)  
- **Batch size:** 32  
- **Epochs:** 10 (baseline only)

#### **Results**
After training the CNN on the training set, the following baseline scores were obtained on the validation set:

| Metric | Score |
|:--------|:------|
| Accuracy | 0.81 |
| Precision | 0.78 |
| Recall | 0.74 |
| F1-score | 0.76 |

*(These results represent the performance of a non-optimized CNN, serving as a reference point.)*
